Population vs sample, parameter vs statistic, and variable scale classification.

```python
import pandas as pd
df = pd.read_csv('../../data/applied-statistics/study_hours_scores.csv')
print(df.head())
print('n =', len(df))
print('Sample mean exam score =', round(df['exam_score'].mean(), 2))
```

```python
scale_map = {
    'study_hours_per_week': 'ratio',
    'attendance_pct': 'ratio',
    'exam_score': 'interval/ratio (context dependent)'
}
for k, v in scale_map.items():
    print(f'{k}: {v}')
```
