# STAT101 Week 9: Descriptive Statistics
Center, spread, percentiles, and outlier diagnostics from real data.

```python
import pandas as pd
df = pd.read_csv('../../data/applied-statistics/manufacturing_defects.csv')
x = df['defects_found']
print('mean=', round(x.mean(), 2))
print('median=', x.median())
print('std=', round(x.std(ddof=1), 2))
print('Q1=', x.quantile(0.25), 'Q3=', x.quantile(0.75))
```

```python
q1, q3 = x.quantile(0.25), x.quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5*iqr
upper = q3 + 1.5*iqr
outliers = df[(x < lower) | (x > upper)]
print('IQR=', iqr)
print('Outliers rows=')
print(outliers[['batch_id', 'defects_found']])
```
