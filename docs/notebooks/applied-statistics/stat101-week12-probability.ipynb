# STAT101 Week 12: Probability and Conditional Probability
Compute empirical probabilities and Bayes-style updates from tabular data.

```python
import pandas as pd
df = pd.read_csv('../../data/applied-statistics/health_trial.csv')
p_side_effect = (df['side_effect'] == 'yes').mean()
p_treat = (df['group'] == 'treatment').mean()
p_side_given_treat = ((df['group'] == 'treatment') & (df['side_effect'] == 'yes')).sum() / (df['group'] == 'treatment').sum()
print('P(side_effect)=', round(p_side_effect, 3))
print('P(treatment)=', round(p_treat, 3))
print('P(side_effect | treatment)=', round(p_side_given_treat, 3))
```

```python
# Quick Bayes update with made-up sensitivity/specificity values
p_disease = 0.08
sens = 0.92
spec = 0.88
p_pos = sens * p_disease + (1-spec) * (1-p_disease)
p_disease_given_pos = sens * p_disease / p_pos
print('P(disease | positive test)=', round(p_disease_given_pos, 3))
```
