```python
import pandas as pd
import numpy as np
df = pd.read_csv('../../data/applied-statistics/study_hours_scores.csv')
print(df.describe())
corr = df['study_hours_per_week'].corr(df['exam_score'])
print('Correlation(hours, score)=', round(corr, 3))
```

```python
# Simple regression by formulas
x = df['study_hours_per_week']
y = df['exam_score']
xbar, ybar = x.mean(), y.mean()
b1 = ((x-xbar)*(y-ybar)).sum() / ((x-xbar)**2).sum()
b0 = ybar - b1*xbar
yhat = b0 + b1*x
resid = y - yhat
r2 = 1 - (resid**2).sum() / ((y-ybar)**2).sum()
print(f'Model: score_hat = {b0:.3f} + {b1:.3f}*hours')
print(f'R^2 = {r2:.3f}')
```
