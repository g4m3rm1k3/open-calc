# STAT101 Week 3: Confidence Intervals
Build 95% confidence intervals for means and proportions from CSV data.

```python
import pandas as pd
import numpy as np
df = pd.read_csv('../../data/applied-statistics/study_hours_scores.csv')
x = df['exam_score']
n = len(x)
xbar = x.mean()
s = x.std(ddof=1)
se = s / np.sqrt(n)
me = 1.96 * se
print(f'95% CI mean score: [{xbar-me:.2f}, {xbar+me:.2f}]')
```

```python
# Proportion CI example: score >= 75
p_hat = (df['exam_score'] >= 75).mean()
se_p = np.sqrt(p_hat * (1 - p_hat) / n)
me_p = 1.96 * se_p
print(f'p-hat={p_hat:.3f}, 95% CI=[{p_hat-me_p:.3f}, {p_hat+me_p:.3f}]')
```
